```python
# Lat.5
from abc import ABC, abstractmethod

#1. Ini adalah KONTRAK. Semua turunan wajib punya method di bawah ini.
class GameUnit(ABC):
  @abstractmethod
  def serang(self, target):
    pass
  @abstractmethod
  def info(self):
    pass

#2. Implementasi pada Class Konkret
class Hero(GameUnit):
  def __init__(self, nama):
    self.nama = nama
  # Kita WAJIB membuat method serang, kalau tidak akan Error
  def serang(self, target):
    print(f"Hero {self.nama} menebas {target}!")
  def info(self):  print(f"Saya adalah Hero: {self.nama}")

class Monster(GameUnit):
  def __init__(self, jenis):
    self.jenis = jenis
  # Implementasi serang versi Monster
  def serang(self, target):
    print(f"Monster {self.jenis} menggigit {target}!")
  def info(self):
    print(f"Saya adalah Monster: {self.jenis}")

# -- Uji Coba --
#unit = GameUnit()
# ERROR! Abstract class tidak bisa jadi objek.
h = Hero("Alucard")
m = Monster("Serigala")
h.info()
h.serang("Sebalak")
m.info()
m.serang("Alucard")

#--------------
# Analisa
#--------------
# 1. dikarenakan kita sudah menjanjikan beberapa fitur, sehingga kita harus mengerjakan apa yang sudah kita janjikan atau tawarkan
# 2. Dikarenakan Game Unit sudah menjadi parameter sehingga tidak boleh dijadikan objek, nanti error.
```

```python
# Lat.6
# Parent Class
class Hero:
  def __init__(self, nama):
    self.nama = nama
  def serang(self):
    print(f"Hero menyerang dengan tangan kosong")

# Child CLass 1
class Mage(Hero) :
  def serang(self) :
    print (f"{self.nama} (Mage) menembakkan Listrik! Boom!")

# Child Class 2
class Archer(Hero):
  def tembak_panah(self):
    print(f"{self.nama} (Archer) memanah dari jauh! Jleb!")

# Child Class 3
class Fighter(Hero):
  def serang(self):
    print(f"{self.nama} (Fighter) memukul dengan pedang! Slash!")
# Child Class New
class Healer(Hero) :
  def serang(self) :
    print (f"{self.nama} tidak menyerang, tapi menyembuhkan teman!")

# -- Penerapan Polymorphism --
# Kita punya daftar hero campuran
pasukan = [
 Mage("Lisa"),
 Archer("Amber"),
 Fighter("Kazuha"),
 Fighter("Razor"),
 Healer("Barbara")
]

print("--- PERANG DIMULAI ---")
# Satu perintah loop, tapi respon berbeda-beda (Polymorphism)
for pahlawan in pasukan:
 pahlawan.serang()

#--------------
# Analisa
#--------------
# 1. berjalan lancar. Polymorphism memungkinkan kita untuk menambahkan suatu pembaruan.
# 2. tidak ada perubahan. Harus persis sama karena agar program berjalan.
```

```python
# Tugas Project ver2

from abc import ABC, abstractmethod


# ==========================================
# 1. ABSTRACTION (KERANGKA)
# ==========================================

class KamarHotel(ABC):

    def __init__(kamar, nama_kamar, harga_dasar):
        kamar.nama_kamar = nama_kamar
        kamar.__stok = 50
        kamar.__harga_dasar = harga_dasar

    # GETTER: melihat stok kamar
    def get_stok(kamar):
        return kamar.__stok

    # GETTER: melihat harga dasar
    def get_harga_dasar(kamar):
        return kamar.__harga_dasar

    # Method untuk memesan kamar
    def pesan_kamar(kamar, jumlah):
        if jumlah <= 0:
            print("Jumlah kamar harus lebih dari 0.")
            return False

        elif jumlah > kamar.__stok:
            print(
                f"Maaf, {kamar.nama_kamar} penuh! "
                f"Stok tersedia hanya {kamar.__stok} kamar."
            )
            return False

        else:
            kamar.__stok -= jumlah
            print(f"Berhasil memesan {jumlah} kamar.")
            print(f"Sisa stok: {kamar.__stok} kamar.")
            return True

    # SETTER: mengubah stok kamar
    def tambah_stok(kamar, jumlah):
        if jumlah < 0:
            print(
                f"Gagal update stok {kamar.nama_kamar}! "
                f"Stok tidak boleh negatif ({jumlah})."
            )
        else:
            kamar.__stok += jumlah

    # ABSTRACT METHOD
    @abstractmethod
    def tampilkan_detail(kamar):
        pass

    # ABSTRACT METHOD
    @abstractmethod
    def hitung_harga_total(kamar, jumlah_kamar, jumlah_malam):
        pass


# ==========================================
# 2. INHERITANCE + POLYMORPHISM
# ==========================================

class KamarDeluxe(KamarHotel):

    def __init__(kamar, nama_kamar, harga_dasar, fasilitas):
        super().__init__(nama_kamar, harga_dasar)
        kamar.fasilitas = fasilitas

    def tampilkan_detail(kamar):
        print(
            f"[DELUXE] {kamar.nama_kamar} | "
            f"Fasilitas: {kamar.fasilitas}"
        )

    def hitung_harga_total(kamar, jumlah_kamar, jumlah_malam):
        harga = kamar.get_harga_dasar()
        pajak = harga * 0.10

        subtotal = (
            (harga + pajak)
            * jumlah_kamar
            * jumlah_malam
        )

        return subtotal


class KamarStandard(KamarHotel):

    def __init__(kamar, nama_kamar, harga_dasar, kapasitas):
        super().__init__(nama_kamar, harga_dasar)
        kamar.kapasitas = kapasitas

    def tampilkan_detail(kamar):
        print(
            f"[STANDARD] {kamar.nama_kamar} | "
            f"Kapasitas: {kamar.kapasitas}"
        )

    def hitung_harga_total(kamar, jumlah_kamar, jumlah_malam):
        harga = kamar.get_harga_dasar()
        pajak = harga * 0.05

        subtotal = (
            (harga + pajak)
            * jumlah_kamar
            * jumlah_malam
        )

        return subtotal


# ==========================================
# 3. FUNCTION TRANSAKSI
# ==========================================

def proses_transaksi(daftar_pesanan):

    total_tagihan = 0

    print("\n--- STRUK PEMESANAN ---")

    for nomor, pesanan in enumerate(daftar_pesanan, start=1):

        kamar = pesanan["kamar"]
        jumlah_kamar = pesanan["jumlah_kamar"]
        jumlah_malam = pesanan["malam"]

        print(f"\n{nomor}. ", end="")
        kamar.tampilkan_detail()

        print(f"Jumlah kamar: {jumlah_kamar}")
        print(f"Jumlah malam: {jumlah_malam}")

        subtotal = kamar.hitung_harga_total(
            jumlah_kamar,
            jumlah_malam
        )

        print(f"Subtotal: Rp {subtotal:,.0f}")

        total_tagihan += subtotal

    print("\n----------------------------------------")
    print(f"TOTAL TAGIHAN: Rp {total_tagihan:,.0f}")
    print("----------------------------------------")


# ==========================================
# 4. SETUP DATA KAMAR
# ==========================================

print("--- SETUP DATA KAMAR ---")

# Membuat kamar Deluxe
kamar_deluxe = KamarDeluxe(
    "Kamar Deluxe Sea View",
    1_500_000,
    "Private Pool"
)

# Membuat kamar Standard
kamar_standard = KamarStandard(
    "Kamar Standard Superior",
    500_000,
    "2 Orang"
)


# ==========================================
# 5. MENAMPILKAN STOK AWAL
# ==========================================

print("\n--- KETERSEDIAAN KAMAR ---")

print(
    f"{kamar_deluxe.nama_kamar} : "
    f"{kamar_deluxe.get_stok()} kamar"
)

print(
    f"{kamar_standard.nama_kamar} : "
    f"{kamar_standard.get_stok()} kamar"
)


# ==========================================
# 6. PEMESANAN
# ==========================================

print("\n--- PEMESANAN ---")

daftar_pesanan = []


# ------------------------------------------
# PEMESANAN DELUXE
# ------------------------------------------

while True:
    try:
        jumlah_deluxe = int(
            input("Berapa kamar Deluxe yang dipesan? ")
        )

        if jumlah_deluxe < 0:
            print("Masukkan angka yang benar!")
            continue

        break

    except ValueError:
        print("Masukkan angka yang benar!")


if jumlah_deluxe > 0:

    if kamar_deluxe.pesan_kamar(jumlah_deluxe):

        while True:
            try:
                malam_deluxe = int(
                    input("Berapa malam menginap? ")
                )

                if malam_deluxe <= 0:
                    print("Masukkan angka yang benar!")
                    continue

                break

            except ValueError:
                print("Masukkan angka yang benar!")

        daftar_pesanan.append({
            "kamar": kamar_deluxe,
            "jumlah_kamar": jumlah_deluxe,
            "malam": malam_deluxe
        })


# ------------------------------------------
# PEMESANAN STANDARD
# ------------------------------------------

while True:
    try:
        jumlah_standard = int(
            input("Berapa kamar Standard yang dipesan? ")
        )

        if jumlah_standard < 0:
            print("Masukkan angka yang benar!")
            continue

        break

    except ValueError:
        print("Masukkan angka yang benar!")


if jumlah_standard > 0:

    if kamar_standard.pesan_kamar(jumlah_standard):

        while True:
            try:
                malam_standard = int(
                    input("Berapa malam menginap? ")
                )

                if malam_standard <= 0:
                    print("Masukkan angka yang benar!")
                    continue

                break

            except ValueError:
                print("Masukkan angka yang benar!")

        daftar_pesanan.append({
            "kamar": kamar_standard,
            "jumlah_kamar": jumlah_standard,
            "malam": malam_standard
        })


# ==========================================
# PROSES TRANSAKSI
# ==========================================

proses_transaksi(daftar_pesanan)
```
